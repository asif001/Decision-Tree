```python
# Question 1: What is a Decision Tree, and how does it work in the context of classification?
''' Answer:
A Decision Tree is a supervised machine learning algorithm used for both classification and regression problems.
It represents decisions in the form of a tree-like structure consisting of a root node, internal decision nodes, branches, and leaf nodes.

How a Decision Tree works:
1 Start with the root node:
The complete training dataset is initially placed at the root.
2 Select the best feature:
The algorithm evaluates the available features and chooses the feature that provides the best separation of the target classes.
Measures such as Gini Impurity, Entropy, or Information Gain can be used.
3 Split the data:
The dataset is divided into smaller groups based on the selected feature and its value.
4 Repeat the process:
The same process is recursively applied to the resulting subsets.
5 Stop splitting:
Splitting stops when a stopping condition is reached, such as a maximum tree depth, minimum number of samples, or when the node becomes sufficiently pure.
6 Make predictions:
The final nodes are called leaf nodes. Each leaf represents a predicted class.
```

```python
# Question 2  Explain the concepts of Gini Impurity and Entropy as impurity measures. How do they impact the splits in a Decision Tree?
''' Answer:
1. Gini Impurity:-
Gini Impurity measures the probability that a randomly selected observation would be incorrectly classified if it were assigned a class according to the distribution of classes in the node.
2. Entropy :- Entropy measures the uncertainty or randomness in a node.

Impact on Decision Tree Splits
The Decision Tree considers different possible splits and selects the split that produces the greatest reduction in impurity.
For entropy, this reduction is called Information Gain.
For Gini, the tree generally chooses the split that produces the greatest reduction in Gini impurity.
```

```python
# Question 3  What is the difference between Pre-Pruning and Post-Pruning in Decision Trees? Give one practical advantage of using each.
''' Answer:
Pre-Pruning:-
Pre-Pruning stops the tree from growing further during the training proces
Practical advantage: It reduces training time and prevents the tree from becoming unnecessarily complex.

Post-Pruning:-
Post-Pruning first allows the tree to grow and then removes branches that do not provide sufficient predictive value
Practical advantage:
It can remove unnecessary branches after examining the complete tree, often improving generalization on unseen data.
```

```python
# Question 4  What is Information Gain in Decision Trees, and why is it important for choosing the best split?
''' Answer:
Information Gain measures how much uncertainty or impurity is reduced after splitting a dataset according to a particular feature.
Why is Information Gain important?
A Decision Tree may have many possible features and many possible split points.
Information Gain provides a systematic way to determine which split gives the greatest improvement in class separation.
```

```python
# Question 5 What are some common real-world applications of Decision Trees, and what are their main advantages and limitations?
''' Anser:
Decision Trees are widely used because their decision-making process is relatively easy to understand.
Advantages
Easy to interpret: The decision-making process can be represented visually.
Little preprocessing: Feature scaling is generally not required.
Handles nonlinear relationships: Trees can model complex relationships between features and target variables.
Works with numerical and categorical information: Depending on the implementation and preprocessing.
Useful for both classification and regression.

Limitations
Overfitting: Deep trees can memorize training data.
Instability: Small changes in the training data can result in a substantially different tree.
Greedy splitting: The locally best split may not always produce the globally best tree.
Large trees can become difficult to interpret.
Single trees may have lower predictive performance than ensemble methods such as Random Forests or Gradient Boosting in many applications.
```

```python
# Question 6: Write a Python program to load the Iris Dataset, train a Decision Tree Classifier using the Gini criterion, and print accuracy and feature importances.
# Answer:
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# Load Iris dataset
iris = load_iris()

X = iris.data
y = iris.target

# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Create Decision Tree Classifier using Gini criterion
model = DecisionTreeClassifier(
    criterion="gini",
    random_state=42
)

# Train the model
model.fit(X_train, y_train)

# Make predictions
y_pred = model.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

# Print results
print("Accuracy:", accuracy)

print("\nFeature Importances:")
for feature, importance in zip(iris.feature_names,
                               model.feature_importances_):
    print(f"{feature}: {importance:.4f}")
```

```python
# Question 7: Train a Decision Tree Classifier with max_depth=3 and compare its accuracy to a fully-grown tree.
# Answer:
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# Load Iris dataset
iris = load_iris()

X = iris.data
y = iris.target

# Split data
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Tree with maximum depth = 3
tree_depth_3 = DecisionTreeClassifier(
    max_depth=3,
    random_state=42
)

tree_depth_3.fit(X_train, y_train)

pred_depth_3 = tree_depth_3.predict(X_test)

accuracy_depth_3 = accuracy_score(
    y_test,
    pred_depth_3
)

# Fully grown tree
fully_grown_tree = DecisionTreeClassifier(
    random_state=42
)

fully_grown_tree.fit(X_train, y_train)

pred_full = fully_grown_tree.predict(X_test)

accuracy_full = accuracy_score(
    y_test,
    pred_full
)

# Display results
print("Accuracy with max_depth=3:",
      accuracy_depth_3)

print("Accuracy of fully-grown tree:",
      accuracy_full)
```

```python
# Load the Boston Housing Dataset, train a Decision Tree Regressor, and print MSE and feature importances.
# Answer:
from sklearn.datasets import fetch_openml
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error
import pandas as pd

# Load Boston Housing dataset from OpenML
boston = fetch_openml(
    name="boston",
    version=1,
    as_frame=True
)

X = boston.data
y = pd.to_numeric(boston.target)

# Convert all features to numeric values
X = X.apply(pd.to_numeric)

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Create Decision Tree Regressor
model = DecisionTreeRegressor(
    random_state=42
)

# Train model
model.fit(X_train, y_train)

# Predict
y_pred = model.predict(X_test)

# Calculate MSE
mse = mean_squared_error(y_test, y_pred)

print("Mean Squared Error:", mse)

print("\nFeature Importances:")

for feature, importance in zip(
    X.columns,
    model.feature_importances_
):
    print(f"{feature}: {importance:.4f}")
```

```python
# Question 9: Tune max_depth and min_samples_split using GridSearchCV and print the best parameters and accuracy.
# Answer:
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# Load Iris dataset
iris = load_iris()

X = iris.data
y = iris.target

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Create Decision Tree
model = DecisionTreeClassifier(
    random_state=42
)

# Define parameter grid
param_grid = {
    "max_depth": [1, 2, 3, 4, 5, None],
    "min_samples_split": [2, 5, 10, 15]
}

# GridSearchCV
grid_search = GridSearchCV(
    estimator=model,
    param_grid=param_grid,
    cv=5,
    scoring="accuracy"
)

# Train and tune model
grid_search.fit(X_train, y_train)

# Best parameters
print("Best Parameters:")
print(grid_search.best_params_)

# Best cross-validation score
print("\nBest Cross-Validation Accuracy:")
print(grid_search.best_score_)

# Evaluate best model on test data
best_model = grid_search.best_estimator_

y_pred = best_model.predict(X_test)

test_accuracy = accuracy_score(
    y_test,
    y_pred
)

print("\nTest Accuracy:")
print(test_accuracy)
```
